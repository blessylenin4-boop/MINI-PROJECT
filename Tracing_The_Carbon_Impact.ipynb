# 📊 Dataset Information

## Dataset Overview

The dataset contains industry-level greenhouse gas (GHG) emission factors for U.S. commodities and industries. It helps us understand and compare the carbon impact associated with different industries.

### Dataset Details

| Attribute                   | Details                                                      |
| --------------------------- | ------------------------------------------------------------ |
| **Dataset Name**            | Supply Chain Greenhouse Gas Emission Factors v1.3 by NAICS-6 |
| **Records**                 | 1,016                                                        |
| **Original Features**       | 8                                                            |
| **Domain**                  | Environmental Data Analytics                                 |
| **Industry Classification** | 2017 NAICS                                                   |
| **GHG Coverage**            | All GHGs                                                     |
| **Main Measurement**        | Supply Chain Emission Factors                                |
| **Unit**                    | kg CO₂e / 2022 USD, purchaser price                          |

## Key Variables

* **2017 NAICS Code** – Industry classification code.
* **2017 NAICS Title** – Name of the industry.
* **GHG** – Greenhouse gas category represented in the dataset.
* **Unit** – Measurement unit used for the emission factors.
* **Supply Chain Emission Factors without Margins** – Emission factor excluding margins.
* **Margins of Supply Chain Emission Factors** – Margin component associated with the emission factor.
* **Supply Chain Emission Factors with Margins** – Emission factor including margins.
* **Reference USEEIO Code** – Reference code associated with the industry.

## Why This Dataset?

This dataset is suitable for exploring how carbon impact varies across industries. The industry classifications and emission factors allow us to perform statistical analysis, comparisons, and meaningful visualizations.

# 📊 Stage 4 — Data Visualization

In this stage, charts and graphs are used to visually represent carbon impact across industries and identify important patterns, comparisons and relationships in the dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10_industries,
    x='Supply Chain Emission Factors with Margins',
    y='2017 NAICS Title'
)

plt.title('Top 10 Industries by Carbon Impact')
plt.xlabel('Supply Chain Emission Factors with Margins')
plt.ylabel('Industry')
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Bottom 10 Industries by Carbon Impact

bottom_10_industries = df.nsmallest(
    10,
    'Supply Chain Emission Factors with Margins'
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=bottom_10_industries,
    x='Supply Chain Emission Factors with Margins',
    y='2017 NAICS Title',
    color='#6BAED6'S
)

plt.title('Bottom 10 Industries by Carbon Impact', fontsize=16, fontweight='bold')
plt.xlabel('Supply Chain Emission Factors with Margins (kg CO₂e / USD)')
plt.ylabel('Industry')
plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Distribution of Carbon Impact

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Supply Chain Emission Factors with Margins',
    bins=30,
    kde=True,
    color='#2E8B57'
)

plt.title('Distribution of Carbon Impact Across Industries',
          fontsize=16, fontweight='bold')
plt.xlabel('Supply Chain Emission Factors with Margins (kg CO₂e / USD)')
plt.ylabel('Number of Industries')
plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Carbon Impact Without vs With Margins

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Supply Chain Emission Factors without Margins',
    y='Supply Chain Emission Factors with Margins',
    color='#8E44AD',
    alpha=0.7,
    s=70
)

plt.title('Carbon Impact: Without vs With Margins',
          fontsize=16, fontweight='bold')
plt.xlabel('Emission Factor Without Margins (kg CO₂e / USD)')
plt.ylabel('Emission Factor With Margins (kg CO₂e / USD)')
plt.tight_layout()
plt.show()
```

```python
# Visualization 5: Carbon Impact Share of Top 5 Industries

top_5 = df.nlargest(
    5,
    'Supply Chain Emission Factors with Margins'
)

plt.figure(figsize=(8, 8))

plt.pie(
    top_5['Supply Chain Emission Factors with Margins'],
    labels=top_5['2017 NAICS Title'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title(
    'Carbon Impact Share of Top 5 Industries',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# Visualization 6: Distribution and Outliers in Carbon Impact

plt.figure(figsize=(10, 6))

sns.boxplot(
    y=df['Supply Chain Emission Factors with Margins'],
    color='#66C2A5'
)

plt.title(
    'Distribution and Outliers of Carbon Impact',
    fontsize=16,
    fontweight='bold'
)

plt.ylabel('Emission Factor with Margins (kg CO₂e / USD)')

plt.tight_layout()
plt.show()
```

```python
# Visualization 7: Correlation Heatmap

correlation_data = df[
    [
        'Supply Chain Emission Factors without Margins',
        'Margins of Supply Chain Emission Factors',
        'Supply Chain Emission Factors with Margins',
        'Margin Contribution',
        'Margin Contribution %'
    ]
]

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu',
    linewidths=0.5
)

plt.title(
    'Correlation Between Carbon Impact Measures',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# Visualization 8: 'Composition of Summed Emission Factors',

total_without = df['Supply Chain Emission Factors without Margins'].sum()
total_margin = df['Margin Contribution'].sum()

values = [total_without, total_margin]
labels = ['Without Margins', 'Margin Contribution']

plt.figure(figsize=(8, 8))

plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.45)
)

plt.title(
    'Composition of Summed Emission Factors',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# Visualization 9: Industries with Highest Margin Contribution

top_margin = df.nlargest(
    10,
    'Margin Contribution'
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_margin['2017 NAICS Title'],
    top_margin['Margin Contribution']
)

plt.title(
    'Top 10 Industries by Margin Contribution',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Margin Contribution (kg CO₂e / USD)')
plt.ylabel('Industry')

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Visualization 10: Carbon Impact Distribution by Measure

plot_data = df[
    [
        'Supply Chain Emission Factors without Margins',
        'Margins of Supply Chain Emission Factors',
        'Supply Chain Emission Factors with Margins'
    ]
]

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=plot_data,
    inner='quartile'
)

plt.title(
    'Distribution of Carbon Impact Measures',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Carbon Impact Measure')
plt.ylabel('Emission Factor (kg CO₂e / USD)')

plt.xticks(
    [0, 1, 2],
    ['Without Margins', 'Margins', 'With Margins']
)

plt.tight_layout()
plt.show()
```

```python
# Visualization 11: Carbon Impact Components of Top 5 Industries

top_5_components = df.nlargest(
    5,
    'Supply Chain Emission Factors with Margins'
).copy()

plt.figure(figsize=(11, 6))

plt.bar(
    top_5_components['2017 NAICS Title'],
    top_5_components['Supply Chain Emission Factors without Margins'],
    label='Without Margins'
)

plt.bar(
    top_5_components['2017 NAICS Title'],
    top_5_components['Margin Contribution'],
    bottom=top_5_components['Supply Chain Emission Factors without Margins'],
    label='Margin Contribution'
)

plt.title(
    'Carbon Impact Components of Top 5 Industries',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Industry')
plt.ylabel('Emission Factor (kg CO₂e / USD)')
plt.xticks(rotation=45, ha='right')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Visualization 12: Carbon Impact vs Margin Contribution %

plt.figure(figsize=(11, 6))

scatter = plt.scatter(
    df['Supply Chain Emission Factors with Margins'],
    df['Margin Contribution %'],
    s=df['Margin Contribution'] * 2500 + 20,
    alpha=0.6
)

plt.title(
    'Carbon Impact vs Margin Contribution Percentage',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Emission Factor with Margins (kg CO₂e / USD)')
plt.ylabel('Margin Contribution (%)')
plt.text(0.02, 0.98, 'Bubble size represents margin contribution',
         transform=plt.gca().transAxes, va='top', fontsize=10)
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
# Key Statistical Findings

print("Highest Carbon Impact:")
print(
    df.loc[
        df['Supply Chain Emission Factors with Margins'].idxmax(),
        ['2017 NAICS Title', 'Supply Chain Emission Factors with Margins']
    ]
)

print("\nLowest Carbon Impact:")
print(
    df.loc[
        df['Supply Chain Emission Factors with Margins'].idxmin(),
        ['2017 NAICS Title', 'Supply Chain Emission Factors with Margins']
    ]
)

print("\nAverage Carbon Impact:")
print(
    df['Supply Chain Emission Factors with Margins'].mean()
)

print("\nAverage Margin Contribution:")
print(
    df['Margin Contribution'].mean()
)

print("\nHighest Margin Contribution:")
print(
    df.loc[
        df['Margin Contribution'].idxmax(),
        ['2017 NAICS Title', 'Margin Contribution']
    ]
)
```

# 💡 Stage 5 — Insights & Recommendations

In this stage, the results from the analysis and visualizations are interpreted to identify key findings, patterns and areas for improvement. Recommendations are provided based on the observed carbon impact across industries.

# 💡 Key Insights

### 1. Highest Carbon Impact
Cement Manufacturing has the highest carbon impact in the dataset, with an emission factor of **3.924 kg CO₂e per USD** including margins.

### 2. Lowest Carbon Impact
Insurance Agencies and Brokerages has the lowest carbon impact, with an emission factor of only **0.029 kg CO₂e per USD** including margins.

### 3. Average Carbon Impact
The average carbon impact across all industries is approximately **0.2819 kg CO₂e per USD**.

### 4. Margin Contribution
The average margin contribution is approximately **0.0169 kg CO₂e per USD**, showing that margins add an additional component to the overall supply-chain emission factor.

### 5. Highest Margin Contribution
Construction Sand and Gravel Mining has the highest margin contribution, at **0.125 kg CO₂e per USD**.

### 6. Large Variation Across Industries
The difference between the highest and lowest emission factors shows that carbon impact varies substantially across industries.

### 7. Importance of Industry-Level Analysis
The results show that industry-level analysis is important because different industries can have significantly different carbon impacts and margin contributions.

# 🏁 Conclusion

The analysis of the Supply Chain Greenhouse Gas Emission Factors dataset shows that carbon impact varies significantly across industries. The results identify industries with particularly high emission factors, while some industries have considerably lower carbon impacts.

Cement Manufacturing recorded the highest carbon impact in the dataset, while Insurance Agencies and Brokerages recorded the lowest. The analysis also shows that margins contribute an additional component to overall supply-chain emissions.
